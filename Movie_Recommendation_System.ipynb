**Project Title:** Movie Recommendation System.

**Description:** Built a recommendation engine using Collaborative Filtering (based on user ratings) to suggest films similar to a target movie.

**Tools Used:** Python, Pandas, and Scikit-learn.

**How to Run:** "This project is optimized for Google Colab. Click the 'Open in Colab' button to run the code in your browser."

```python
import pandas as pd
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
print("Environment is ready!")
```

```python
# Download and unzip the dataset directly into Colab
!wget http://files.grouplens.org/datasets/movielens/ml-100k.zip
!unzip ml-100k.zip

# Load the ratings data
ratings_cols = ['user_id', 'movie_id', 'rating', 'timestamp']
ratings = pd.read_csv('ml-100k/u.data', sep='\t', names=ratings_cols)

# Load the movie titles
items_cols = ['movie_id', 'movie_title' , 'release_date', 'video_release_date', 'IMDb_URL', 'unknown', 'Action', 'Adventure', 'Animation', 'Childrens', 'Comedy', 'Crime', 'Documentary', 'Drama', 'Fantasy', 'Film-Noir', 'Horror', 'Musical', 'Mystery', 'Romance', 'Sci-Fi', 'Thriller', 'War', 'Western']
movies = pd.read_csv('ml-100k/u.item', sep='|', names=items_cols, encoding='latin-1')

# Merge them to see titles instead of just IDs
movie_data = pd.merge(ratings, movies[['movie_id', 'movie_title']], on='movie_id')
movie_data.head()
```

```python
# Rows: User IDs | Columns: Movie Titles | Values: Their ratings
movie_matrix = movie_data.pivot_table(index='user_id', columns='movie_title', values='rating')
# Fill missing ratings with 0
movie_matrix.fillna(0, inplace=True)
```

```python
def get_recommendations(target_movie):
    # Get ratings for the target movie
    target_movie_ratings = movie_matrix[target_movie]

    # Calculate correlation with all other movies
    similar_to_target = movie_matrix.corrwith(target_movie_ratings)

    # Clean up the results
    corr_df = pd.DataFrame(similar_to_target, columns=['Correlation'])
    corr_df.dropna(inplace=True)

    # Return the top 5 most similar movies
    return corr_df.sort_values('Correlation', ascending=False).head(6)

# Test it with a movie title from the dataset
print(get_recommendations('Star Wars (1977)'))
```
