```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

# 1. Load a sample dataset (using a public URL for the SMS Spam collection)
url = 'https://raw.githubusercontent.com/justmarkham/pycon-2016-tutorial/master/data/sms.tsv'
df = pd.read_table(url, header=None, names=['label', 'message'])

# 2. Convert labels to numbers (ham=0, spam=1)
df['label_num'] = df.label.map({'ham':0, 'spam':1})

# Change test_test=0.25 to test_size=0.25
X_train, X_test, y_train, y_test = train_test_split(df.message, df.label_num, test_size=0.25)
# 4. Vectorize the text (Turn words into numbers)
vect = CountVectorizer()
X_train_dtm = vect.fit_transform(X_train)
X_test_dtm = vect.transform(X_test)

# 5. Train the Naive Bayes model (Required for Project 7)
nb = MultinomialNB()
nb.fit(X_train_dtm, y_train)

# 6. Test the model
print("Model Accuracy:", nb.score(X_test_dtm, y_test))

# Test with your own message
sample_message = ["Congratulations! You've won a $1000 Walmart gift card. Click here."]
sample_dtm = vect.transform(sample_message)
print("Prediction (1=Spam, 0=Ham):", nb.predict(sample_dtm))
```
